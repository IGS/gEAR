# gEAR Starter Notebook (Python)

This notebook is meant to help you begin exploring the dataset you selected from gEAR.

The flow is intentionally simple:
- prepare the Python environment
- confirm which file was mounted for this session
- load the data only when you are ready

You do not need to run every cell at once. Start with the setup cell, then move to the load cell when you want to inspect the data.

```python
import os
from pathlib import Path

import numpy as np
import pandas as pd
import scanpy as sc
import matplotlib.pyplot as plt

sc.settings.set_figure_params(dpi=120, facecolor="white")
plt.rcParams["figure.figsize"] = (6, 4)

selected_dataset = os.environ.get("GEAR_SELECTED_DATASET", "").strip()
data_root = Path("/data")
target_path = data_root / selected_dataset if selected_dataset else None

print("Dataset label:", selected_dataset or "not provided")
print("Expected file path:", target_path if target_path is not None else "not assigned")

if target_path is None:
    raise RuntimeError("No dataset label was provided for this notebook session.")

if not target_path.exists():
    raise FileNotFoundError(f"The expected dataset file is missing: {target_path}")

print("The file is present and ready for the next step.")
```

```python
# Load the dataset into memory

adata = sc.read_h5ad(target_path)

print("Loaded dataset successfully.")
print("Number of cells:", adata.n_obs)
print("Number of features:", adata.n_vars)

adata
```

```python
# Suggested next steps

print("Observation columns:")
print(list(adata.obs.columns[:10]))

print("\nFeature columns:")
print(list(adata.var.columns[:10]))

print("\nAvailable matrix-style annotations:")
print(list(adata.obsm.keys()) if hasattr(adata, "obsm") else "No obsm entries found")
```

```python
# Optional first plot

if "X_umap" in adata.obsm:
    sc.pl.embedding(adata, basis="umap", show=False)
else:
    print("No umap coordinates were found in adata.obsm.")
```
