# gEAR Starter Notebook (R)

This notebook is meant to help you begin exploring the dataset you selected from gEAR.

The flow is intentionally simple:
- prepare the R environment
- confirm which file was mounted for this session
- load the data only when you are ready

You do not need to run every cell at once. Start with the setup cell, then move to the load cell when you want to inspect the data.

```r
# Setup and path checks

library(zellkonverter)
library(scater)

selected_dataset <- Sys.getenv("GEAR_SELECTED_DATASET", unset = "")
selected_dataset <- trimws(selected_dataset)

data_root <- "/data"
target_path <- if (nzchar(selected_dataset)) file.path(data_root, selected_dataset) else NA_character_

cat("Dataset label:", if (nzchar(selected_dataset)) selected_dataset else "not provided", "\n")
cat("Expected file path:", if (!is.na(target_path)) target_path else "not assigned", "\n")

if (!nzchar(selected_dataset)) {
  stop("No dataset label was provided for this notebook session.")
}

if (!file.exists(target_path)) {
  stop(sprintf("The expected dataset file is missing: %s", target_path))
}

cat("The file is present and ready for the next step.\n")
```

```r
# Load the dataset into memory

sce <- readH5AD(target_path)

print("Loaded dataset successfully.")
print(sprintf("Number of cells: %d", ncol(sce)))
print(sprintf("Number of features: %d", nrow(sce)))

sce
```

```r
# Suggested next steps

cat("Column data fields:\n")
print(colnames(colData(sce)))

cat("\nRow data fields:\n")
print(colnames(rowData(sce)))

cat("\nAvailable reduced dimensions:\n")
print(names(reducedDims(sce)))
```

```r
# Optional first plot

reduced_names <- names(reducedDims(sce))

if (length(reduced_names) > 0) {
  reduced_name <- reduced_names[1]
  cat("Using reduced dimension:", reduced_name, "\n")
  plotReducedDim(sce, dimred = reduced_name)
} else {
  cat("No reduced dimensions were found in the dataset.\n")
}
```
